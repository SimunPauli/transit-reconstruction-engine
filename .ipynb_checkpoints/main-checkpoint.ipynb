```python
import requests

url = "http://localhost:8080/otp/routers/default/index/graphql"

query = """
{
  plan(
    from: {lat: 55.66667064297805, lon: 12.603777629035296}
    to: {lat: 55.78029108132897, lon: 12.516761369213794}
    transportModes: [{mode: WALK}, {mode: TRANSIT}]
  ) {
    itineraries {
      duration
      legs {
        mode
        startTime
        endTime
        from { name }
        to { name }
      }
    }
  }
}
"""
```

```python
response = requests.post(
    url,
    json={"query": query}
)
print(response.json())
```

```python
data = response.json()
print(data)
```
